```python
df["Age"]=df['Age'].fillna(df['Age'].mean())
```

```python
df = df.drop('Cabin',axis=1)
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.isnull().sum()
```

```python
def fun(n):
    if n=="male":
       return 1
    else:
       return 0
```

```python
df["Sex"] = df["Sex"].apply(fun)
```
