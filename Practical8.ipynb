```python
def fun1(value):
    if (value == "male"):
        return 1
    else:
        return 0
```

```python
def fun2(value):
    if (value == 'S'):
        return 0
    elif (value == 'C'):
        return 1
    elif (value == 'Q'):
        return 2
    else:
        return 0
```

```python
df["Sex"] = df["Sex"].apply(fun1)
```

```python
df["Embarked"] = df["Embarked"].apply(fun2)
```

```python

df.isna().sum()
```

```python
df = df.drop("Cabin", axis=1)
```

```python

df.shape
```

```python
df.info()
```

```python
# =========================================
# Count Plot - Survival Count
# =========================================

plt.figure(figsize=(6,5))

sns.countplot(x='Survived',data=df)

plt.title("Survival Count")

plt.show()
```

```python
# =========================================
# Count Plot - Gender Wise Survival
# =========================================

plt.figure(figsize=(6,5))

sns.countplot(
    x='Sex',
    hue='Survived',
    data=df
)

plt.title("Gender Wise Survival")

plt.show()
```

```python
# =========================================
# Histogram of Fare
# =========================================

plt.figure(figsize=(8,5))

plt.hist(
    df['Fare'],
    bins=20
)

plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.title("Distribution of Fare")

plt.show()
```
