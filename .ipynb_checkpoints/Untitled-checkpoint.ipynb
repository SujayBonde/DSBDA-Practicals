```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("weather.txt")
```

```python
df
```

```python
avg_temp = df["Temperature"].mean()
avg_dew = df["DewPoint"].mean()
avg_wind = df["WindSpeed"].mean()
```

```python
print("Average Temperature:", avg_temp)
print("Average Dew Point:", avg_dew)
print("Average Wind Speed:", avg_wind)
```
