* 2. Write a Python program to display some basic statistical details like percentile, mean,
standard deviation etc. of the species of ‘Iris-setosa’, ‘Iris-versicolor’ and ‘Iris-versicolor’
of iris.csv dataset.

```python
iris = pd.read_csv("Iris.csv")
```

```python
df = pd.DataFrame(iris)
df
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
irisGroup= df.groupby('Species')
```

```python
irisGroup
```

```python
print(irisGroup.mean(numeric_only=True))
```

```python
print(irisGroup.median(numeric_only=True))
```

```python
print(irisGroup.std(numeric_only=True))
```

```python
df.describe(percentiles=[0.25, 0.5, 0.75])
```

```python
for name, data in irisGroup:

    print(f"\n==============================")
    print(f"Statistics for {name}")
    print(f"==============================")

    print("\nMean:\n")
    print(data.mean(numeric_only=True))

    print("\nMedian:\n")
    print(data.median(numeric_only=True))

    print("\nStandard Deviation:\n")
    print(data.std(numeric_only=True))

    print("\nPercentiles:\n")
    print(
        data.describe(percentiles=[0.25, 0.5, 0.75])
    )
```
